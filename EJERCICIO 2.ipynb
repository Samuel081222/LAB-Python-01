```python
#3.
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
suma = 0
for t in temperaturas:
  suma = suma + t
promedio = suma / len(temperaturas)
print("El promedio de las temperaturas es:", promedio)
for  t in temperaturas:
  if t > promedio:
    print("temperatura mayo")
```
