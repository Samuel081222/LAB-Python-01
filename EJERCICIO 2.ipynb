```python
#3.
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
suma = 0
for t in temperaturas:
  suma = suma + t
promedio = suma / len(temperaturas)
print("El promedio de las temperaturas es:", promedio)
for  t in temperaturas:
  if t > promedio:
    print("temperatura mayo")
```

```python
#4.
suma_pares = 0
suma_impares = 0

for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

resultado = {
    "suma_pares": suma_pares,
    "suma_impares": suma_impares
}

print("Estado final:", resultado)
```

```python
#5.

productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))


    if precio < 0 or cantidad < 0:
        continue

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("Total del inventario:", total)

if productos:
    mas_caro = productos[0]
    for p in productos:
        if p["precio"] > mas_caro["precio"]:
            mas_caro = p
    print("Producto más caro:", mas_caro["nombre"])

print(productos)


```
