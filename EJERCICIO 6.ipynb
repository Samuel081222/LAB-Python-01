```python
#1.
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]

nombre_nuevo = input("Ingrese el nombre del nuevo producto: ")

precio_valido = False
precio_nuevo = 0
while not precio_valido:
    try:
        precio_nuevo = float(input("Ingrese el precio del producto: "))
    except ValueError:
        print("Error: Debe ingresar un valor numerico valido.")
    else:
        print("Precio registrado exitosamente.")
        precio_valido = True
    finally:
        print("Proceso de lectura de precio finalizado.")

stock_valido = False
stock_nuevo = 0
while not stock_valido:
    try:
        stock_nuevo = int(input("Ingrese el stock del producto: "))
    except ValueError:
        print("Error: Debe ingresar un numero entero valido.")
    else:
        print("Stock registrado exitosamente.")
        stock_valido = True
    finally:
        print("Proceso de lectura de stock finalizado.")

nuevo_producto = {
    "nombre": nombre_nuevo,
    "precio": precio_nuevo,
    "stock": stock_nuevo
}

inventario.append(nuevo_producto)
print("Inventario actualizado:")
print(inventario)
```

else: Este bloque se ejecuta únicamente cuando el bloque try termina de ejecutarse sin haber lanzado ninguna excepción (es decir, el usuario introdujo un dato válido que se pudo convertir correctamente). Permite separar la lógica exitosa del manejo de errores del except.

finally: Este bloque se ejecuta siempre, sin importar si ocurrió un error en el try o si se ejecutó con éxito. Es útil para realizar acciones de limpieza o mostrar mensajes de cierre que deben estar presentes en cada iteración del ciclo

```python
#2.
import json

inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Cafe", "precio": 8000, "stock": 10},
]

archivo = open("inventario.csv", "w")
for producto in inventario:
    linea = producto["nombre"] + "," + str(producto["precio"]) + "," + str(producto["stock"]) + "\n"
    archivo.write(linea)
archivo.close()

def leer_csv():
    resultado = []
    try:
        archivo = open("inventario.csv", "r")
        for linea in archivo:
            linea_limpia = linea.replace("\n", "")
            datos = linea_limpia.split(",")
            producto = {
                "nombre": datos[0],
                "precio": float(datos[1]),
                "stock": int(datos[2])
            }
            resultado.append(producto)
        archivo.close()
    except FileNotFoundError:
        resultado = []
    return resultado

print("Inventario cargado desde CSV:", leer_csv())

archivo_j = open("inventario.json", "w")
json.dump(inventario, archivo_j)
archivo_j.close()

def leer_json():
    try:
        archivo_j = open("inventario.json", "r")
        resultado = json.load(archivo_j)
        archivo_j.close()
    except FileNotFoundError:
        resultado = []
    return resultado

print("Inventario cargado desde JSON:", leer_json())
```

Ventajas de JSON frente a CSV
Tipos de datos nativos: JSON preserva automáticamente los tipos de datos (números enteros, decimales, booleanos y listas), mientras que el formato CSV guarda todo como texto plano, lo que obliga a realizar conversiones manuales como float() o int() al leer los datos.

Estructuras complejas: JSON soporta de forma nativa estructuras anidadas complejas (diccionarios dentro de listas o múltiples niveles de jerarquía), mientras que CSV se limita exclusivamente a un formato tabular plano de filas y columnas.
