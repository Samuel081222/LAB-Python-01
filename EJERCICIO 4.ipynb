```python
#1.A
#CON GLOBAL
contador_global = 0

def incrementar():
    global contador_global
    contador_global = contador_global + 1

incrementar()
print(contador_global)
```

```python
#1.B
#Haciendo que la función reciba el valor y retorne el nuevo valor
contador_global = 0

def incrementar(valor):
    return valor + 1

contador_global = incrementar(contador_global)
print(contador_global)
```

Al ejecutar el código se obtiene el siguiente error:UnboundLocalError: local variable 'contador_global' referenced before assignmenT
Esto ocurre porque Python asume que cualquier variable que se asigne o modifique dentro de una función (contador_global = contador_global + 1) es una variable local. Cuando Python intenta evaluar la parte derecha de la asignación (contador_global + 1), busca una variable local llamada contador_global que aún no ha sido creada ni inicializada en ese ámbito, lo que genera el error.

```python
#2.
def agregar_nota(notas, nueva):
    notas.append(nueva)

def reiniciar(notas):
    notas.clear()

mis_notas = [3.5, 4.0]
print("Id original:", id(mis_notas))

agregar_nota(mis_notas, 4.8)
print("Id despues de agregar:", id(mis_notas))

reiniciar(mis_notas)
print("Id despues de reiniciar:", id(mis_notas))
print("Estado final de la lista:", mis_notas)
```

Predicción del resultado
Al ejecutar el código original se imprime:[3.5, 4.0, 4.8]

Explicación:

La lista comienza con [3.5, 4.0].

La función agregar_nota utiliza el método .append(), el cual modifica la lista original directamente en memoria, agregando el 4.8.

La función reiniciar hace una reasignación local (notas = []). Esto hace que la variable local notas apunte a una nueva lista vacía en la memoria, pero no afecta a la variable original mis_notas, que sigue apuntando al objeto anterior. Por lo tanto, al imprimir mis_notas al final, conserva el valor modificado.

Diferencia entre mutación y reasignación
Modificar un objeto mutable (append, clear): Altera el contenido del objeto existente en la memoria. Como las listas son mutables y se pasan por referencia (o asignación de objetos), cualquier cambio interno repercute en todas las variables que apunten a ese mismo objeto.

Reasignar una variable local (notas = []): Cambia el enlace de la variable local para que apunte a un nuevo objeto en la memoria. El objeto original que venía de afuera queda intacto porque la variable local simplemente se desvincula de él.

```python
#3.
cuenta = {"titular": "Ana", "saldo": 200000, "movimientos": []}

def consultar(cuenta):
    print("Titular:", cuenta["titular"])
    print("Saldo actual:", cuenta["saldo"])

def depositar(cuenta, monto):
    cuenta["saldo"] = cuenta["saldo"] + monto
    movimiento = {"tipo": "Deposito", "monto": monto, "saldo_resultante": cuenta["saldo"]}
    cuenta["movimientos"].append(movimiento)
    print("Deposito exitoso.")

def retirar(cuenta, monto):
    if monto > cuenta["saldo"]:
        print("Fondos insuficientes. No se puede dejar el saldo en negativo.")
    else:
        cuenta["saldo"] = cuenta["saldo"] - monto
        movimiento = {"tipo": "Retiro", "monto": monto, "saldo_resultante": cuenta["saldo"]}
        cuenta["movimientos"].append(movimiento)
        print("Retiro exitoso.")

activo = True
while activo:
    print("\n--- CAJERO AUTOMATICO ---")
    print("1. Consultar saldo")
    print("2. Depositar")
    print("3. Retirar")
    print("4. Salir")

    opcion = input("Ingrese una opcion: ")

    if opcion == "1":
        consultar(cuenta)
    elif opcion == "2":
        monto = float(input("Ingrese el monto a depositar: "))
        depositar(cuenta, monto)
    elif opcion == "3":
        monto = float(input("Ingrese el monto a retirar: "))
        retirar(cuenta, monto)
    elif opcion == "4":
        activo = False
    else:
        print("Opcion no valida.")

print("\n--- EXTRACTO DE MOVIMIENTOS ---")
print("Tipo\t\tMonto\t\tSaldo Resultante")
print("-" * 50)
for mov in cuenta["movimientos"]:
    print(mov["tipo"] + "\t\t" + str(mov["monto"]) + "\t\t" + str(mov["saldo_resultante"]))
```
