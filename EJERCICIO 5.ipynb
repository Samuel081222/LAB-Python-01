```python
#1.
import random

ganadas = 0
perdidas = 0
jugando = True

while jugando:
    secreto = random.randint(1, 100)
    intentos = []
    contador = 0
    adivinado = False

    while contador < 7 and not adivinado:
        intento = int(input("Ingrese un numero entre 1 y 100: "))
        intentos.append(intento)
        contador = contador + 1

        if intento == secreto:
            adivinado = True
            print("Adivinaste el numero")
        elif intento < secreto:
            print("El numero secreto es mayor")
        else:
            print("El numero secreto es menor")

    if adivinado:
        ganadas = ganadas + 1
    else:
        perdidas = perdidas + 1
        print("El numero secreto era", secreto)

    print("Historial de intentos:", intentos)
    print("Partidas ganadas:", ganadas)
    print("Partidas perdidas:", perdidas)

    respuesta = input("Desea jugar otra vez (si/no): ")
    if respuesta == "no":
        jugando = False
```

Los 7 intentos son suficientes porque se fundamentan en el principio de la búsqueda binaria. Cada vez que adivinas un número intermedio, descartas exactamente la mitad del rango restante (por ejemplo, si el rango es de 1 a 100 y adivinas 50, en un solo intento reduces el espacio de búsqueda a 50 números posibles).

```python
#2.
def crear_tablero():
    tablero = []
    for i in range(10):
        fila = []
        for j in range(10):
            fila.append(0)
        tablero.append(fila)
    return tablero

def contar_vecinos(tablero, x, y):
    contador = 0
    for dx in range(-1, 2):
        for dy in range(-1, 2):
            if dx != 0 or dy != 0:
                nx = x + dx
                ny = y + dy
                if nx >= 0 and nx < 10 and ny >= 0 and ny < 10:
                    if tablero[nx][ny] == 1:
                        contador = contador + 1
    return contador

def siguiente_generacion(tablero):
    nuevo_tablero = []
    for i in range(10):
        fila = []
        for j in range(10):
            fila.append(0)
        nuevo_tablero.append(fila)

    for i in range(10):
        for j in range(10):
            vecinos = contar_vecinos(tablero, i, j)
            if tablero[i][j] == 1:
                if vecinos == 2 or vecinos == 3:
                    nuevo_tablero[i][j] = 1
                else:
                    nuevo_tablero[i][j] = 0
            else:
                if vecinos == 3:
                    nuevo_tablero[i][j] = 1
                else:
                    nuevo_tablero[i][j] = 0
    return nuevo_tablero

def mostrar_tablero(tablero):
    for fila in tablero:
        linea = ""
        for celda in fila:
            if celda == 1:
                linea = linea + "■ "
            else:
                linea = linea + "· "
        print(linea)
    print("-" * 20)

tablero = crear_tablero()
tablero[4][4] = 1
tablero[4][5] = 1
tablero[4][6] = 1

for gen in range(5):
    print("Generacion", gen + 1)
    mostrar_tablero(tablero)
    tablero = siguiente_generacion(tablero)
```

Es necesario porque las reglas del Juego de la Vida exigen que todas las celdas cambien de estado de forma simultánea (en un solo paso de tiempo discreto).

Si modificaras el tablero actual directamente mientras lo vas recorriendo, las celdas que ya fueron evaluadas y actualizadas (por ejemplo, celdas que mueren o nacen antes en el bucle) alterarían de inmediato el conteo de vecinos para las celdas vecinas que todavía no han sido evaluadas en esa misma generación. Esto corrompería la lógica matemática del autómata celular, haciendo que los resultados dependan del orden de recorrido de los bucles en lugar de reflejar el estado global previo.
